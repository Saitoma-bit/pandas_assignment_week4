```python
# Part 3: Clean the Data

# Perform appropriate cleaning steps:
    # 1. Handle missing values: decide whether to fill or drop them.
#handling misisng values in Quantity colunm

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

df["Quantity"]
   
            
```

```python
# Perform appropriate cleaning steps:
    # 1. Handle missing values: decide whether to fill or drop them.

#handling misisng values in Unit_Price colunm

df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())

df["Unit_Price"]
   
```

```python
# 2. Remove duplicate rows.

df["Product_ID"].to_list()

df["Product_ID"]
```

```python
# 2. Remove duplicate rows.

df = df.drop_duplicates()

df
```

```python
# 3. Standardize text: make all Product_Name, Category, and Supplier lowercase (or title case).
# Product_Name to lowercase or title case.

df["Product_Name"].to_list()
    
  
```

```python
df["Category"] = df["Category"].str.title()

df["Category"]
```

```python
# Standardize text: make all Product_Name, Category, and Supplier lowercase (or title case).
# Supplier to lowercase or title case.

df["Supplier"].to_list()
```

```python
# 4. Fix data types: convert columns to proper numeric or datetime types.
# for Quantity
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

df["Quantity"]
   
```

```python
# 4. Fix data types: convert columns to proper numeric or datetime types.
# for Unit_Price
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")

df["Unit_Price"]
   
```

```python
# 4. Fix data types: convert columns to proper numeric or datetime types.
# for Reorder_Level
df["Reorder_Level"] = pd.to_numeric(df["Reorder_Level"], errors="coerce")

df["Reorder_Level"]
   
```

```python
#5. Handle incorrect values:
           # Replace or drop rows with negative or unrealistic prices/quantities.
# to replace use .fillana 
# to drop use .dropna
```

```python
# Standardize date formats in Last_Updated.

df["Last_Update"].to_list()
```

```python
# Standardize date formats in Last_Updated.

def date_parser(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT
df["Last_Update"] = df["Last_Update"].apply(date_parser)

df["Last_Update"]
```

```python
# Part 4: Analyze and Sort the Data
# Use Pandas to answer the following: 
# 1. Which products need restocking (where Quantity < Reorder_Level)? 
# Products needing restock

restock_products = df[df["Quantity"] < df["Reorder_Level"]]

print("Products Needing Restock:")
print(restock_products[["Product_Name", "Quantity", "Reorder_Level"]])

```

```python
# 2. What is the average unit price per category?
# Average Unit Price per Category

avg_price = df.groupby("Category")["Unit_Price"].mean()

print("\nAverage Unit Price Per Category:")
print(avg_price)

```

```python
# 3. Which supplier provides the most products? 
# Supplier with Most Products

supplier_count = df["Supplier"].value_counts()

print("\nSupplier Providing Most Products:")
print(supplier_count)

```

```python
# 4. List the top 5 most expensive products. 
# Top 5 Most Expensive Products

top_expensive = df.nlargest(5, "Unit_Price")

print("\nTop 5 Most Expensive Products:")
print(top_expensive[["Product_Name", "Unit_Price"]])

```

```python
# 5. Sort the cleaned dataset by Category and then by Unit_Price in descending order.
# Sort by Category and Unit Price

sorted_df = df.sort_values(
    by=["Category", "Unit_Price"],
    ascending=[True, False]
)

print("\nSorted Dataset:")
print(sorted_df)
 
```
